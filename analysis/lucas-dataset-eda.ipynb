```python
import pandas as pd
import geopandas as gpd

# load lucas merged dataset
gdf = gpd.read_file('D:\\tierra\\data\\Lucas\\merged_lucas_dataset.geojson')

pd.set_option('display.max_columns', None)  # Show all columns in the DataFrame
gdf.head()
```

```python
# missing values in the dataset
missing_values = gdf.isnull().sum()
print("Missing values in the dataset:")
print(missing_values[missing_values > 0])

# plot missing values
def plot_missing_values(gdf):
    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.figure(figsize=(12, 6))
    sns.heatmap(gdf.isnull(), cbar=False, cmap='viridis')
    plt.title('Missing Values Heatmap')
    plt.show()

plot_missing_values(gdf)
```

```python
# remove columns with more than 50% missing values
def remove_high_missing_columns(gdf, threshold=0.5):
    missing_percentage = gdf.isnull().mean()
    columns_to_drop = missing_percentage[missing_percentage > threshold].index
    gdf_cleaned = gdf.drop(columns=columns_to_drop)
    return gdf_cleaned

gdf_cleaned = remove_high_missing_columns(gdf, threshold=0.5)

# remove rows with any missing values
gdf_cleaned = gdf_cleaned.dropna()
print("Rows removed due to missing values:", gdf.shape[0] - gdf_cleaned.shape[0])

print(gdf_cleaned.columns)

print("Cleaned DataFrame shape:", gdf_cleaned.shape)

plot_missing_values(gdf_cleaned)
```
