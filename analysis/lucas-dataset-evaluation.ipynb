# Curating Datasets

```python
import pandas as pd
import geopandas as gpd

base_path = 'D:\\tierra\\data\\Lucas\\'

def load_file(file_path):
    # check file extension
    if file_path.endswith('.shp'):
        # Load shapefile using geopandas
        df = gpd.read_file(base_path + file_path)
    elif file_path.endswith('.geojson'):
        # Load GeoJSON file using geopandas
        df = gpd.read_file(base_path + file_path)
    elif file_path.endswith('.csv'):
        # Load CSV file using pandas
        df = pd.read_csv(base_path + file_path)
    else:
        raise ValueError("Unsupported file format: {}".format(file_path))

    return df
def merge_point_id(df1, df2, key1='POINTID', key2='POINTID'):
    if key1 not in df1.columns or key2 not in df2.columns:
        raise ValueError("Both DataFrames must contain a '{}' column.".format(key1))

    merged_df = pd.merge(df1, df2, left_on=key1, right_on=key2, how='outer')
    return merged_df
```

```python
point_idx = load_file('soil\\LUCAS-SOIL-2018-v2\\LUCAS-SOIL-2018.shp')
point_idx.head()
```

```python
soil_df = load_file('soil\\LUCAS-SOIL-2018-v2\\LUCAS-SOIL-2018.csv')
print(soil_df.columns)
```

```python
bd_2 = load_file('soil\\LUCAS-SOIL-2018-v2\\BulkDensity_2018_final-2.csv')
bd_2.head()
```

```python
merged_df = merge_point_id(soil_df, bd_2, key1='POINTID', key2='POINT_ID')
# drop
merged_df.drop(columns=['POINT_ID'], inplace=True)
merged_df.head()
```

```python
org = load_file('soil\\LUCAS-SOIL-2018-v2\\LUCAS2018_ORG.csv')
org.head()
```

```python
merged_df = merge_point_id(merged_df, org, key1='POINTID', key2='POINT_ID')
# drop
merged_df.drop(columns=['POINT_ID'], inplace=True)
merged_df.head()
```

```python
erosion = load_file('soil\\LUCAS-SOIL-2018-v2\\LUCAS2018_EROSION.csv')
erosion.head()
```

```python
merged_df = merge_point_id(merged_df, erosion, key1='POINTID', key2='POINT_ID')
# drop
merged_df.drop(columns=['POINT_ID'], inplace=True)
merged_df.head()
```

```python
bd_df_paper = load_file('bulk_density\\BD2018_data_for_paper.csv')
bd_df_paper.head()
```

```python
merged_df = merge_point_id(merged_df, bd_df_paper, key1='POINTID', key2='POINT_ID')
# drop
merged_df.drop(columns=['POINT_ID'], inplace=True)
merged_df.head()
```

```python
bd_shp = load_file('bulk_density\\BD2018_data_for_paper_shapefile\\BD2018_fef_landcover.shp')
bd_shp.head()
```

```python
bd_shp.drop(columns=['geometry'], inplace=True)
merged_df = merge_point_id(merged_df, bd_shp, key1='POINTID', key2='POINT_ID')
# drop
merged_df.drop(columns=['POINT_ID'], inplace=True)
merged_df.head()
```

```python
texture_shp = load_file('texture\\LUCAS_texture_all\\LUCAS_Texture_all.shp')
texture_shp.head()
```

```python
texture_shp.drop(columns=['geometry'], inplace=True)
merged_df = merge_point_id(merged_df, texture_shp)
merged_df.head()
```

# Final Dataset

```python
print("Final merged DataFrame shape:", merged_df.shape)
print(merged_df.columns.tolist())

gdf = merge_point_id(merged_df, point_idx)

# convert to GeoDataFrame
gdf = gpd.GeoDataFrame(gdf, geometry='geometry', crs='EPSG:4326')

gdf.head()
```

```python
# save as geojson
geojson_path = base_path + 'merged_lucas_dataset.geojson'
gdf.to_file(geojson_path, driver='GeoJSON')
print(f"GeoJSON file saved to: {geojson_path}")

# save as CSV
csv_path = base_path + 'merged_lucas_dataset.csv'
gdf.to_csv(csv_path, index=False)
print(f"CSV file saved to: {csv_path}")
```
